```python
# 1. Section: Exploratory Data Analysis with Visualizations

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the churn dataset
churn_df = pd.read_csv('../data/customer_churn.csv')
print(churn_df.head())
print('\nDataset shape:', churn_df.shape)
print('\nMissing values:\n', churn_df.isnull().sum())
print('\nChurn rate:', churn_df['Churn'].mean())

# Summary statistics
print(churn_df.describe(include='all').T)

# Distribution of churn target
plt.figure(figsize=(6, 4))
sns.countplot(x='Churn', data=churn_df, palette='Set2')
plt.title('Churn Distribution')
plt.show()

# Monthly charges by churn status
plt.figure(figsize=(8, 5))
sns.boxplot(x='Churn', y='MonthlyCharges', data=churn_df, palette='Set2')
plt.title('Monthly Charges vs Churn')
plt.show()

# Tenure vs churn
plt.figure(figsize=(8, 5))
sns.boxplot(x='Churn', y='TenureMonths', data=churn_df, palette='Set2')
plt.title('Tenure vs Churn')
plt.show()

# Correlation heatmap
numeric_df = churn_df.select_dtypes(include=['number'])
plt.figure(figsize=(10, 8))
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap')
plt.show()

# Key findings
print('''
Key findings from EDA:
- Customers with shorter tenure are more likely to churn.
- Higher monthly charges usually correlate with churn risk.
- Contract type and payment method strongly influence churn behavior.
- Customers with limited support and lower service engagement are more likely to leave.
''')
```
